```python
import time
import configparser
import requests
import pandas as pd

def CrmBitrix(content='', params=[], result='', pause_duration=60, max_retries=3, batch_limit=30000):
    config = configparser.ConfigParser()
    config.read('Config.ini')

    if 'BITRIX' not in config:
        raise ValueError('Section BITRIX manquante dans le fichier \'config.ini\'')

    BITRIX_DOMAIN = config.get('BITRIX', 'BITRIX_DOMAIN')
    BITRIX_TOKEN = config.get('BITRIX', 'BITRIX_TOKEN')

    if not BITRIX_DOMAIN or not BITRIX_TOKEN:
        raise ValueError('Domaine et/ou token Bitrix manquants dans le fichier \'config.ini\'')

    params_str = '&'.join(params)
    BitrixUrl = f'https://{BITRIX_DOMAIN}/rest/1/{BITRIX_TOKEN}/{content}?{params_str}&start='

    listId, listTITLE, listEmails, listNumeros, listDATE_CREATE = [], [], [], [], []
    listActivite1, listActivite2, listActivite3 = [], [], []
    listAdresse, listCodePostale, listVille, listSiret, listidd, listphones = [], [], [], [], [], []

    start = 0
    total_records = 0

    while True:
        url = BitrixUrl + str(start)
        
        for attempt in range(max_retries):
            try:
                response = requests.get(url)
                response.raise_for_status()
                break  
            except requests.exceptions.HTTPError as e:
                if response.status_code == 429:
                    wait_time = pause_duration * (2 ** attempt)
                    print(f'Rate limit exceeded. Retrying in {wait_time} seconds...')
                    time.sleep(wait_time)
                else:
                    raise ValueError(f'Erreur lors de la récupération depuis Bitrix : {e}')
            except requests.exceptions.RequestException as e:
                raise ValueError(f'Erreur lors de la récupération depuis Bitrix : {e}')

        data = response.json()

        if 'result' in data:
            contacts = data['result']
            for contact in contacts:
                listId.append(contact.get('ID', ''))
                listTITLE.append(contact.get('TITLE', ''))
                listEmails.append(contact.get('EMAIL', [{}])[0].get('VALUE', ''))
                listNumeros.append(contact.get('PHONE', [{}])[0].get('VALUE', ''))
                listphones.append(contact.get('UF_CRM_1741080141', ''))
                listDATE_CREATE.append(contact.get('DATE_CREATE', ''))
                listActivite1.append(contact.get('UF_CRM_1686295056', ''))
                listActivite2.append(contact.get('UF_CRM_1686295079', ''))
                listActivite3.append(contact.get('UF_CRM_1686295100', ''))
                listAdresse.append(contact.get('UF_CRM_1686295131', ''))
                listCodePostale.append(contact.get('UF_CRM_1686295169', ''))
                listVille.append(contact.get('UF_CRM_1686295194', ''))
                listSiret.append(contact.get('UF_CRM_1686295022', ''))
                listidd.append(contact.get('UF_CRM_1722346888', ''))
                

            total_records += len(contacts) 

            if total_records >= batch_limit:
                print(f"Atteint {batch_limit} enregistrements, pause de 5 minutes...")
                time.sleep(300) 
                total_records = 0 

        if 'next' in data:
            start = data['next']
        else:
            break

    df_btx = pd.DataFrame({
        'ID': listId,
        'CompanyId': listId,
        'nom': listTITLE,
        "Email": listEmails,
        "Crée": listDATE_CREATE,
        "Mobile": listNumeros,
        "Phones": listphones,
        "Activite1": listActivite1,
        "Activite2": listActivite2,
        "Activite3": listActivite3,
        "Adresse": listAdresse,
        "CodePostale": listCodePostale,
        "Ville": listVille,
        "Siret": listSiret,
        "IDS": listidd
    })

    return df_btx

dfb = CrmBitrix(
    content='crm.company.list.json',
    params=[
        'select[]=ID',
        'select[]=TITLE',
        'select[]=EMAIL',
        'select[]=PHONE',
        'select[]=UF_CRM_1741080141',
        'select[]=DATE_CREATE',
        'select[]=UF_CRM_1686295056',
        'select[]=UF_CRM_1686295079',
        'select[]=UF_CRM_1686295100',
        'select[]=UF_CRM_1686295131',
        'select[]=UF_CRM_1686295169',
        'select[]=UF_CRM_1686295194', 
        'select[]=UF_CRM_1686295022',
        'select[]=UF_CRM_1722346888',
    ],
    result="title"
)
dfb
```

```python
import pandas as pd
import requests
import configparser
import time
import numpy as np
import random

def CrmBitrix(content='', params=[], pause_duration=1, max_retries=3):
    """
    Récupère des entreprises déjà présentes dans Bitrix24 pour éviter les doublons.
    """
    print("Début de la récupération des données de Bitrix24...")

    config = configparser.ConfigParser()
    config.read('config.ini')

    if 'BITRIX' not in config:
        raise ValueError("Section BITRIX manquante dans le fichier 'config.ini'")

    BITRIX_DOMAIN = config.get('BITRIX', 'BITRIX_DOMAIN')
    BITRIX_TOKEN = config.get('BITRIX', 'BITRIX_TOKEN')

    if not BITRIX_DOMAIN or not BITRIX_TOKEN:
        raise ValueError("Domaine et/ou token Bitrix manquants dans le fichier 'config.ini'")

    print(f"Utilisation du domaine Bitrix: {BITRIX_DOMAIN}")

    params_str = '&'.join(params)
    BitrixUrl = f'https://{BITRIX_DOMAIN}/rest/1/{BITRIX_TOKEN}/{content}?{params_str}&start='

    listNoms, listIds, listSIRETs = [], [], []

    start = 0
    while True:
        url = BitrixUrl + str(start)
        print(f"Envoi de la requête à l'URL: {url}")

        for attempt in range(max_retries):
            try:
                response = requests.get(url)
                response.raise_for_status()
                break
            except requests.exceptions.HTTPError as e:
                if response.status_code == 429:
                    wait_time = pause_duration * (2 ** attempt)
                    print(f"Limite atteinte. Nouvelle tentative dans {wait_time} secondes...")
                    time.sleep(wait_time)
                else:
                    raise ValueError(f"Erreur Bitrix: {e}")
            except requests.exceptions.RequestException as e:
                raise ValueError(f"Erreur de connexion: {e}")

        data = response.json()
        if 'result' in data:
            contacts = data['result']
            for contact in contacts:
                listNoms.append(contact.get('TITLE', ''))
                listIds.append(contact.get('ID', ''))
                listSIRETs.append(contact.get('UF_CRM_1686295022', ''))

        if 'next' in data:
            start = data['next']
        else:
            break

    df_btx = pd.DataFrame({'ID': listIds, 'nom': listNoms, 'siretetablissementdeclarant': listSIRETs})
    print("Récupération terminée.")
    return df_btx

def get_statut_value(statut):
    statut_mapping = {
        'Pas Qualiopi': '140',
        'Qualiopi': '142',
        'Qualiopi perdu': '146'
    }
    return statut_mapping.get(statut, '140')

def clean_data(data):
    for key, value in data.items():
        if isinstance(value, (float, np.float64)) and (np.isnan(value) or np.isinf(value)):
            data[key] = ""
    return data

def create_bitrix_company(df_local, df_bitrix, batch_size=50, max_retries=5):
    """
    Crée des entreprises dans Bitrix24 en évitant les doublons et en gérant les erreurs.
    """
    print("Début de la création des entreprises dans Bitrix24...")

    config = configparser.ConfigParser()
    config.read('config.ini')

    BITRIX_DOMAIN = config.get('BITRIX', 'BITRIX_DOMAIN')
    BITRIX_TOKEN = config.get('BITRIX', 'BITRIX_TOKEN')

    if not BITRIX_DOMAIN or not BITRIX_TOKEN:
        raise ValueError("Domaine et/ou token Bitrix manquants dans le fichier 'config.ini'")

    total_created = 0 

    for index, row in df_local.iterrows():
        nom = row.get('denomination', '')
        siret = row.get('siretetablissementdeclarant', '')

        if not df_bitrix[df_bitrix['siretetablissementdeclarant'] == siret].empty:
            print(f"L'entreprise {nom} avec SIRET {siret} existe déjà. Ignorée.")
            continue

        company_data = {
            'TITLE': nom,
            'UF_CRM_1686295022': siret,
            'UF_CRM_1686295131': row.get('adressephysiqueorganismeformation_voie', ''),
            'UF_CRM_1686295169': row.get('adressephysiqueorganismeformation_codepostal', ''),
            'UF_CRM_1686295194': row.get('adressephysiqueorganismeformation_ville', ''),
            'UF_CRM_1686295056': row.get('informationsdeclarees_specialitesdeformation_libellespecialite1', ''),
            'UF_CRM_1722346888': get_statut_value(row.get('statut', ''))
        }

        company_data = clean_data(company_data)

        create_url = f'https://{BITRIX_DOMAIN}/rest/1/{BITRIX_TOKEN}/crm.company.add.json'

        for attempt in range(max_retries):
            response = requests.post(create_url, json={'fields': company_data})

            if response.status_code == 200:
                print(f"Entreprise créée avec succès pour {nom}")
                total_created += 1
                break 
            elif response.status_code == 429:
                wait_time = 2 * (2 ** attempt)
                print(f"Trop de requêtes. Réessai dans {wait_time} secondes...")
                time.sleep(wait_time)
            else:
                print(f"Erreur lors de la création de {nom}: {response.text}")
                break 

        if total_created % batch_size == 0 and total_created > 0:
            pause_time = random.uniform(300, 600) 
            print(f"Pause de {pause_time:.2f} secondes après {batch_size} créations...")
            time.sleep(pause_time)

    print("Création des entreprises terminée.")

input_csv = 'A_SIRET_absents_companyyav.csv'
df_local = pd.read_csv(input_csv, low_memory=False)

df_local.fillna({'informationsdeclarees_nbstagiaires': 0, 'informationsdeclarees_effectifformateurs': 0, 'informationsdeclarees_nbstagiairesconfiesparunautreof': 0}, inplace=True)

df_btx = CrmBitrix(content='crm.company.list.json', params=['select[]=ID', 'select[]=TITLE', 'select[]=UF_CRM_1686295022'])

create_bitrix_company(df_local, df_btx, batch_size=50)

print("Script terminé avec succès !")
```

```python
import pandas as pd
import requests
import configparser
import time
import logging

logging.basicConfig(filename="bitrix_update.log", level=logging.INFO, format="%(asctime)s - %(levelname)s - %(message)s")
console = logging.StreamHandler()
console.setLevel(logging.INFO)
logging.getLogger("").addHandler(console)

logging.info("Démarrage du script")

config = configparser.ConfigParser()
config.read('config.ini')

BITRIX_DOMAIN = config.get('BITRIX', 'BITRIX_DOMAIN')
BITRIX_TOKEN = config.get('BITRIX', 'BITRIX_TOKEN')

if not BITRIX_DOMAIN or not BITRIX_TOKEN:
    logging.error("Domaine et/ou token Bitrix manquants dans config.ini")
    raise ValueError("Domaine et/ou token Bitrix manquants dans config.ini")

csv_file = "entreprises_comgdt.csv"
df_csv = pd.read_csv(csv_file, dtype=str).fillna("")
logging.info(f"CSV chargé : {len(df_csv)} lignes.")

def get_existing_ids_from_bitrix():
    bitrix_url = f"https://{BITRIX_DOMAIN}/rest/1/{BITRIX_TOKEN}/crm.company.list.json"
    params = {"select[]": "ID"}
    existing_ids = set()
    start = 0
    max_retries = 5

    logging.info("Récupération des ID depuis Bitrix...")

    while True:
        for attempt in range(max_retries):
            try:
                response = requests.get(bitrix_url, params={**params, "start": start})
                response.raise_for_status()
                data = response.json()
                break
            except requests.exceptions.RequestException as e:
                wait_time = 10 * (attempt + 1)
                logging.warning(f"Erreur {e}. Nouvelle tentative dans {wait_time} secondes...")
                time.sleep(wait_time)
        else:
            logging.error("Échec après plusieurs tentatives.")
            return existing_ids

        if "result" in data:
            for company in data["result"]:
                existing_ids.add(company["ID"])
        if "next" in data:
            start = data["next"]
        else:
            break

    logging.info(f"{len(existing_ids)} entreprises récupérées de Bitrix.")
    return existing_ids

existing_ids_bitrix = get_existing_ids_from_bitrix()

df_csv_filtered = df_csv[df_csv["ID"].isin(existing_ids_bitrix)]

if df_csv_filtered.empty:
    logging.info("Aucune entreprise du CSV ne correspond à un ID existant dans Bitrix. Fin du script.")
else:
    logging.info(f"{len(df_csv_filtered)} entreprises à mettre à jour.")

    update_url = f"https://{BITRIX_DOMAIN}/rest/1/{BITRIX_TOKEN}/crm.company.update.json"

    max_requests = 300
    pause_duration = 100
    max_retries = 5

    for index, row in df_csv_filtered.iterrows():
        company_id = row["ID"]
       
        uf_value = row["Phones"]

        params = {
            "id": company_id,
            "fields[UF_CRM_1741080141]": uf_value
        }

        for attempt in range(max_retries):
            try:
                response = requests.post(update_url, params=params)
                response.raise_for_status()
                result = response.json()

                if "error" in result:
                    logging.warning(f"rreur pour ID {company_id} : {result['error_description']}")
                else:
                    logging.info(f"ID {company_id} mis à jour avec succès.")
                break
            except requests.exceptions.RequestException as e:
                wait_time = 10 * (attempt + 1)
                logging.warning(f"Erreur réseau pour ID {company_id} : {e}. Nouvelle tentative dans {wait_time} secondes...")
                time.sleep(wait_time)
        else:
            logging.error(f"Échec après {max_retries} tentatives pour ID {company_id}.")

        if (index + 1) % max_requests == 0:
            logging.info(f"Pause de {pause_duration} secondes pour éviter le rate limit")
            time.sleep(pause_duration)

    logging.info("Mise à jour des entreprises terminée.")
```
